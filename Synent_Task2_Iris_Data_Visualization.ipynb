## 1. Load Iris Dataset

```python
import pandas as pd
import matplotlib.pyplot as plt

print("Task 2 - Iris Data Visualization")
```

## 2. Load and Prepare Dataset

```python
from sklearn.datasets import load_iris

iris = load_iris()

df = pd.DataFrame(
    iris.data,
    columns=iris.feature_names
)

df["species"] = iris.target_names[iris.target]

print(df.head())
```

## 3. Bar Chart

```python
avg_petal_length = df.groupby("species")["petal length (cm)"].mean()

avg_petal_length.plot(kind="bar")

plt.title("Average Petal Length by Species")
plt.xlabel("Species")
plt.ylabel("Average Petal Length (cm)")
plt.xticks(rotation=0)
plt.show()
```

## 4. Histogram

```python
plt.hist(df["petal length (cm)"], bins=10)

plt.title("Distribution of Petal Length")
plt.xlabel("Petal Length (cm)")
plt.ylabel("Frequency")
plt.show()
```

## 5. Scatter Plot

```python
plt.scatter(df["petal length (cm)"], df["petal width (cm)"])

plt.title("Petal Length vs Petal Width")
plt.xlabel("Petal Length (cm)")
plt.ylabel("Petal Width (cm)")
plt.show()
```

## 6. Feature Comparison

```python
print(df.groupby("species")[[
    "sepal length (cm)",
    "sepal width (cm)",
    "petal length (cm)",
    "petal width (cm)"
]].mean())
```

## Key Findings

- The Iris dataset contains three species: setosa, versicolor, and virginica.
- Average petal length differs clearly between the three species.
- Setosa has the smallest average petal length and petal width.
- Virginica has the largest average petal length and petal width.
- Versicolor generally has intermediate petal measurements.
- The scatter plot shows a clear relationship between petal length and petal width.

## Conclusion

The Iris dataset shows clear differences in flower measurements across the three species. Petal length and petal width provide useful features for distinguishing the species. Bar charts, histograms, and scatter plots helped visualize these patterns clearly.
