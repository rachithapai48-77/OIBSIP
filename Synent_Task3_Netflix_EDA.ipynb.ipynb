# Task 3 - Netflix Exploratory Data Analysis (EDA)

## 1. Load Netflix Dataset

```python
import pandas as pd
import matplotlib.pyplot as plt

url = "https://raw.githubusercontent.com/rfordatascience/tidytuesday/main/data/2021/2021-04-20/netflix_titles.csv"

df = pd.read_csv(url)

print(df.head())
```

## 2. Dataset Overview

```python
print("Shape:", df.shape)
print("\nColumns:")
print(df.columns)

print("\nDataset Info:")
print(df.info())
```

## 3. Summary Statistics

```python
print("Summary Statistics:")
print(df.describe(include="all"))
```

## 4. Movies vs TV Shows

```python
type_counts = df["type"].value_counts()

print(type_counts)
```

```python
import matplotlib.pyplot as plt

type_counts.plot(kind="bar")

plt.title("Movies vs TV Shows on Netflix")
plt.xlabel("Type")
plt.ylabel("Number of Titles")
plt.xticks(rotation=0)
plt.show()
```

## 5. Release Year Analysis

```python
year_counts = df["release_year"].value_counts().sort_index()

plt.figure(figsize=(12, 5))
plt.plot(year_counts.index, year_counts.values)

plt.title("Netflix Titles by Release Year")
plt.xlabel("Release Year")
plt.ylabel("Number of Titles")
plt.grid()
plt.show()
```

## 6. Correlation Analysis

```python
print("Correlation Matrix:")
print(df.corr(numeric_only=True))
```

## 7. Netflix Titles Added by Year

```python
date_counts = pd.to_datetime(df["date_added"], errors="coerce").dt.year.value_counts().sort_index()

print(date_counts)
```

```python
plt.figure(figsize=(10, 5))
plt.plot(date_counts.index, date_counts.values, marker="o")

plt.title("Netflix Titles Added by Year")
plt.xlabel("Year")
plt.ylabel("Number of Titles Added")
plt.grid()
plt.show()
```

## 8. Rating Analysis

```python
rating_counts = df["rating"].value_counts()

print(rating_counts)
```

```python
rating_counts.plot(kind="bar", figsize=(10, 5))

plt.title("Netflix Titles by Rating")
plt.xlabel("Rating")
plt.ylabel("Number of Titles")
plt.xticks(rotation=45)
plt.show()
```

## Key Findings

- Movies are more common than TV Shows in the dataset.
- The number of Netflix titles added increased significantly over the years.
- 2019 had the highest number of titles added, with 2,136 titles.
- TV-MA is the most common rating, with 2,863 titles.
- The dataset contains only one numeric feature (`release_year`), so no meaningful pairwise correlation can be calculated.

## Conclusion

The exploratory data analysis shows that Netflix has a larger collection of Movies than TV Shows. The number of titles added increased considerably over time, with the highest number recorded in 2019. TV-MA is the most common rating in the dataset. The visualizations and summary statistics help identify important patterns and trends in the Netflix catalog.
